```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
datasets = [
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-2526.csv", "season": "25/26"},
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-2425.csv", "season": "24/25"},
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-2324.csv", "season": "23/24"},
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-2223.csv", "season": "22/23"},
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-2122.csv", "season": "21/22"},
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-2021.csv", "season": "20/21"},
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-1920.csv", "season": "19/20"}, 
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-1819.csv", "season": "18/19"},
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-1718.csv", "season": "17/18"},
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-1617.csv", "season": "16/17"},
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-1516.csv", "season": "15/16"},
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-1415.csv", "season": "14/15"},
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-1314.csv", "season": "13/14"},
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-1213.csv", "season": "12/13"},
    {"file": "https://datahub.io/football/english-premier-league/_r/-/season-1112.csv", "season": "11/12"},
]

df_list = []
for i in datasets:
    df_prem = pd.read_csv(i["file"])
    df_prem['season'] = i['season']
    df_list.append(df_prem)

df = pd.concat(df_list, ignore_index=True)
```

```python
df_merge = df.copy()

col_int = df_merge.select_dtypes(include=['int64']).columns
df_merge[col_int] = df_merge[col_int].astype('int8')

df_merge['Date'] = pd.to_datetime(df_merge['Date']).dt.date
df_merge = df_merge.sort_values('Date', ascending=False)
```
