```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_parquet(r'C:\Users\Quang\Documents\Code\Self Project\PL Prediction\dataset\pl_combined.parquet').sort_values('Date', ascending=True)
```

```python
df.describe().style.format("{:.2f}")
```

```python
df2 = df.copy()
df2["Date"] = pd.to_datetime(df2['Date'])

fig, ax = plt.subplots(figsize=(10,5))
sns.scatterplot(df2[(df2['HomeTeam'] == "Man United" )][['Date', 'FTHG']], x='Date', y='FTHG')
ax.set_xlabel("")
ax.set_ylabel("")
ax.set_title("Duration Distribution")
fig.tight_layout()
plt.show()
```

```python
df2 = df.copy()

fig, ax = plt.subplots(figsize=(3,5))
df2['FTHG'].plot(kind='box', showmeans=True, meanline=True)
ax.set_xlabel("")
ax.set_ylabel("")
ax.set_title("Duration Distribution")
fig.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(10,5))
sns.histplot(df2['FTHG'], bins=30, kde=True)
ax.set_xlabel("")
ax.set_ylabel("")
ax.set_title("Duration Distribution")
fig.tight_layout()
plt.show()
```
